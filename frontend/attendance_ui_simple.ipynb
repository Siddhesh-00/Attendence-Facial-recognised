```python
# Load dependencies
import os
import sys

# Set working directory to project root
os.chdir('/Users/siddhesh/Downloads/PythonProject')
sys.path.insert(0, '/Users/siddhesh/Downloads/PythonProject')

import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

from register import StudentRegistration
from attendance import AttendanceMarker
from trainer import train_from_dataset
from utils import load_encodings, ensure_directories

ensure_directories()
print(f"✅ Working directory: {os.getcwd()}")
print("✅ All modules loaded")
```

```python
# Create output areas
output_area = widgets.Output()
status_area = widgets.Output()

def update_info():
    encodings_data = load_encodings()
    return f"<p><strong>📊 Registered Students:</strong> {len(encodings_data['names'])}</p>"

info_label = widgets.HTML(value=update_info())
print("✅ Output areas created")
```

```python
# Define backend functions
def register_student(name, roll, cls):
    with status_area:
        clear_output()
        try:
            reg = StudentRegistration()
            v, m = reg.validate_input(name, roll, cls)
            if not v:
                print(f"❌ {m}")
                return
            print(f"📸 Capturing faces...")
            s, m = reg.capture_faces(name, roll, cls)
            if not s:
                print(f"❌ {m}")
                return
            print(f"🔧 Generating encodings...")
            s, m = reg.generate_encodings(name, roll, cls)
            if s:
                print(f"✅ Registration successful!")
                info_label.value = update_info()
            else:
                print(f"❌ {m}")
        except Exception as e:
            print(f"❌ Error: {e}")

def start_attendance():
    with status_area:
        clear_output()
        try:
            print("📸 Starting attendance...")
            attendance = AttendanceMarker()
            attendance.run_attendance()
            print(f"✅ Attendance session completed")
        except Exception as e:
            print(f"❌ Error: {e}")

def train_system():
    with status_area:
        clear_output()
        try:
            print("🔄 Training...")
            train_from_dataset()
            encodings_data = load_encodings()
            print(f"✅ Training complete! {len(encodings_data['names'])} students")
            info_label.value = update_info()
        except Exception as e:
            print(f"❌ Error: {e}")

print("✅ Functions defined")
```

```python
# Create buttons and form inputs
register_btn = widgets.Button(description='📝 Register Student', button_style='info', layout=widgets.Layout(width='250px', height='50px'))
attendance_btn = widgets.Button(description='📸 Take Attendance', button_style='success', layout=widgets.Layout(width='250px', height='50px'))
train_btn = widgets.Button(description='🔄 Train Encodings', button_style='warning', layout=widgets.Layout(width='250px', height='50px'))
back_btn = widgets.Button(description='🔙 Back', button_style='primary', layout=widgets.Layout(width='250px', height='40px'))
register_submit = widgets.Button(description='✅ Start', button_style='success', layout=widgets.Layout(width='250px'))

name_input = widgets.Text(placeholder='Name', description='Name:', layout=widgets.Layout(width='400px'))
roll_input = widgets.Text(placeholder='Roll', description='Roll:', layout=widgets.Layout(width='400px'))
class_input = widgets.Text(placeholder='Class', description='Class:', layout=widgets.Layout(width='400px'))

print("✅ Buttons created")
```

```python
# Define navigation functions
def show_menu():
    with output_area:
        clear_output()
        display(widgets.VBox([widgets.HTML("<h2>📋 Main Menu</h2>"), register_btn, attendance_btn, train_btn, info_label]))

def show_register_form():
    with output_area:
        clear_output()
        display(widgets.VBox([widgets.HTML("<h2>📝 Register</h2>"), name_input, roll_input, class_input, register_submit, back_btn]))

register_btn.on_click(lambda b: show_register_form())
attendance_btn.on_click(lambda b: (show_menu.__call__() if False else None, start_attendance()))
train_btn.on_click(lambda b: (show_menu.__call__() if False else None, train_system()))
back_btn.on_click(lambda b: show_menu())
register_submit.on_click(lambda b: (register_student(name_input.value, roll_input.value, class_input.value), 
                                     [setattr(name_input, 'value', ''), setattr(roll_input, 'value', ''), setattr(class_input, 'value', '')]))

print("✅ Navigation ready")
```

```python
# Display main UI
title = widgets.HTML("<h1 style='text-align: center; color: #2c3e50;'>🎓 Smart Face Recognition Attendance System</h1>")
subtitle = widgets.HTML("<h3 style='text-align: center;'>With Blink-Based Anti-Spoofing</h3><hr>")
info_box = widgets.HTML("<div style='background: #ecf0f1; padding: 15px; border-radius: 5px;'><strong>ℹ️ Dataset:</strong> dataset/ | <strong>Reports:</strong> attendance/attendance.xlsx</div>")

main_layout = widgets.VBox([title, subtitle, output_area, status_area, info_box])
display(main_layout)

show_menu()
print("\n✅ System Ready! Click buttons above to start.")
```
